Inspect the Cityscapes dataset

```python
import collections
import logging

from semi_sup_seg.constants import CITYSCAPES_DIR
from semi_sup_seg.data.cityscapes import CityscapesLabeledDataset

logging.basicConfig(level=logging.INFO, force=True)
logger = logging.getLogger(__name__)

%load_ext autoreload
%autoreload 2
```

```python
IS_OOS_SPLITS = ("train", "val")

fine_dataset = CityscapesLabeledDataset(
    label_dir=CITYSCAPES_DIR / "gtFine", selected_is_oos_splits=IS_OOS_SPLITS
)
coarse_dataset = CityscapesLabeledDataset(
    label_dir=CITYSCAPES_DIR / "gtCoarse", selected_is_oos_splits=IS_OOS_SPLITS
)
```

```python
# Inspect the datapoints.

for name, dataset in [
    ("Fine-granularity datapoint index", fine_dataset),
    ("Coarse-granularity datapoint index", coarse_dataset),
]:
    print(f"{name}\n")

    for is_oos_split, datapoints in dataset.datapoint_index.items():
        print(f"\t{is_oos_split}:")

        city_to_datapoints = collections.defaultdict(list)
        for dp in datapoints:
            city_to_datapoints[dp.city].append(dp)

        for city, city_datapoints in sorted(
            city_to_datapoints.items(), key=lambda item: -len(item[1])
        ):
            clips = {dp.clip_idx for dp in city_datapoints}
            print(
                f"\t\t{city:20}: "
                f"{len(city_datapoints):7,} labeled images from "
                f"{len(clips):7,} clips"
            )

        print("\t\t--------------------------------------------")
        print(f"\t\t{'total':20}: {len(datapoints):7,} labeled images")
        print()
```

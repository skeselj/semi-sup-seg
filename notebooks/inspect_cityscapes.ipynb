Inspect the Cityscapes dataset

```python
import logging

from semi_sup_seg.data import CityscapesLabeledDataset

logging.basicConfig(level=logging.INFO, force=True)
logger = logging.getLogger(__name__)

%load_ext autoreload
%autoreload 2
```

```python
dataset = CityscapesLabeledDataset()
```

```python
# Inspect the datapoints.

for name, index in [
    ("Fine-granularity datapoint index", dataset.fine_datapoint_index),
    ("Coarse-granularity datapoint index", dataset.coarse_datapoint_index),
]:
    print(f"{name}\n")

    for is_oos_split in dataset.IS_OOS_SPLITS:
        print(f"\t{is_oos_split}:")
        split_label_count = 0

        for city in sorted(
            index[is_oos_split].keys(),
            key=(
                lambda city: (
                    -sum(len(dps) for dps in index[is_oos_split][city].values())
                )
            ),
        ):
            clip_idx_to_datapoints = index[is_oos_split][city]
            clips = set(clip_idx_to_datapoints.keys())
            frames = {
                (clip_idx, datapoint.label.frame_idx)
                for clip_idx, datapoints in clip_idx_to_datapoints.items()
                for datapoint in datapoints
            }

            split_label_count += len(frames)

            print(
                f"\t\t{city:20}: "
                f"{len(frames):7,} labeled images from {len(clips):7,} clips"
            )

        print("\t\t--------------------------------------------")
        print(f"\t\t{'total':20}: {split_label_count:7,} labeled images")
        print()
```
